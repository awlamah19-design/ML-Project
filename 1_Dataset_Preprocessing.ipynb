```python
# NOTEBOOK 1: DATASET PREPROCESSING
# Student Performance Prediction - Data Preparation

import pandas as pd
import numpy as np

# 1. LOAD DATASET
print("=" * 50)
print("STEP 1: LOADING DATASET")
print("=" * 50)

df = pd.read_csv("data.csv", sep=';')
print(f"Dataset loaded successfully!")
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")

# 2. DISPLAY FIRST FEW ROWS
print("\n" + "=" * 50)
print("STEP 2: FIRST 5 ROWS")
print("=" * 50)
df.head()

# 3. CHECK FOR MISSING VALUES
print("\n" + "=" * 50)
print("STEP 3: CHECKING MISSING VALUES")
print("=" * 50)
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values[missing_values > 0] if any(missing_values > 0) else "No missing values found!")

# 4. CHECK FOR DUPLICATES
print("\n" + "=" * 50)
print("STEP 4: CHECKING DUPLICATES")
print("=" * 50)
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Duplicates removed. New shape: {df.shape}")

# 5. DATA SUMMARY
print("\n" + "=" * 50)
print("STEP 5: DATA SUMMARY STATISTICS")
print("=" * 50)
df.describe(include="all")

# 6. CHECK TARGET VARIABLE DISTRIBUTION
print("\n" + "=" * 50)
print("STEP 6: TARGET VARIABLE DISTRIBUTION")
print("=" * 50)
target_counts = df["Target"].value_counts()
print(target_counts)
print(f"\nPercentages:")
print(df["Target"].value_counts(normalize=True) * 100)

# 7. DATA TYPES
print("\n" + "=" * 50)
print("STEP 7: DATA TYPES")
print("=" * 50)
print(df.dtypes)

# 8. SAVE CLEANED DATA (OPTIONAL)
# df.to_csv("cleaned_data.csv", index=False)
# print("\nCleaned data saved to cleaned_data.csv")

print("\n" + "=" * 50)
print("PREPROCESSING COMPLETE!")
print("=" * 50)
```
